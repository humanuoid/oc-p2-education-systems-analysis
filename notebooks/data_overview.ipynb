# Aperçu des données EdStats

Chaque fichier CSV du dossier `data/input/` est présenté en deux parties :

1. **l'extrait** : les 5 premières lignes du fichier ;
2. **le dictionnaire de données** : nom de chaque colonne (anglais) → définition (français).

Les valeurs manquantes (NaN) sont affichées '-'.

````python
import pandas as pd
from pathlib import Path

INPUT_DIR = Path("data") / "input"
if not INPUT_DIR.exists():  # notebook ouvert depuis le dossier notebooks/
    INPUT_DIR = Path("..") / "data" / "input"


def load_csv(name):
    """Lit le CSV `name` du dossier INPUT_DIR.

    - usecols : on ne garde que les colonnes nommées (la virgule finale de
      chaque ligne du CSV crée sinon une colonne vide "Unnamed").
    - les NaN sont remplacés par '-' : astype(object) passe d'abord les
      colonnes en type "objet" générique, pour que '-' soit accepté même
      dans les colonnes numériques (float).
    """
    df = pd.read_csv(
        INPUT_DIR / name,
        encoding="utf-8",
        usecols=lambda column: "Unnamed" not in column,
    )
    return df.astype(object).fillna("-")


def show_data_dictionary(df, definitions, columns=None):
    """Affiche le dictionnaire de données de `df`, une ligne par colonne.

    - columns : colonnes à afficher (None = toutes les colonnes de df).
    - definitions[col] lève une KeyError si une définition manque.
    - max_colwidth / max_rows à None : aucune troncature à l'affichage.
    """
    if columns is None:
        columns = df.columns
    data_dictionary = pd.DataFrame({
        "Colonne": columns,
        "Définition": [definitions[col] for col in columns],
    })
    with pd.option_context("display.max_colwidth", None, "display.max_rows", None):
        display(data_dictionary)
````

## Mesures par pays et indicateur — EdStatsData.csv

*Valeurs par année : mesures annuelles (1970 à 2017), projections tous les 5 ans (2020 à 2100). Une valeur est une projection quand le nom de l'indicateur contient « Projection » (colonne Indicator Name ; 308 indicateurs sur 3 665, ex. « Wittgenstein Projection: … »).*

### Extrait du CSV

````python
df = load_csv("EdStatsData.csv")
df.head()
````

### Dictionnaire de données

````python
definitions = {
    "Country Name": "Nom du pays ou de l'agrégat régional (ex : 'Arab World') concerné par la ligne.",
    "Country Code": "Code à 3 lettres : ISO3 pour les pays, code interne Banque mondiale pour les agrégats (ARB, EUU…) — clé de jointure avec EdStatsCountry.csv.",
    "Indicator Name": "Nom de l'indicateur mesuré (ex : 'Population, total'). Une projection se repère ici : le nom contient « Projection » (308 indicateurs sur 3 665, ex. « Wittgenstein Projection: … »).",
    "Indicator Code": "Code de l'indicateur (ex : 'SP.POP.TOTL') — clé de jointure avec EdStatsSeries.csv.",
}
# les colonnes d'années (1970 ... 2100) ont toutes la même définition :
# on n'en affiche que les 3 premières et les 3 dernières
annees = [col for col in df.columns if col.isdigit()]  # col.isdigit() = "le nom est un nombre"
for annee in annees[:3] + annees[-3:]:
    definitions[annee] = f"Valeur de l'indicateur pour l'année {annee}"
definitions["..."] = f"... ({len(annees) - 6} années intermédiaires non affichées)"

colonnes = [col for col in df.columns if not col.isdigit()] + annees[:3] + ["..."] + annees[-3:]
show_data_dictionary(df, definitions, columns=colonnes)
````

## Schéma des interactions entre les jeux de données (jointures FK → PK)

Un jeu de données par fichier CSV. Chaque flèche va d'une clé étrangère (FK) vers la clé primaire (PK) à laquelle elle se joint — c'est exactement le `merge` pandas correspondant.

```mermaid
---
title: Jointures entre les jeux de données EdStats (FK vers PK)
---
flowchart LR
    subgraph ref["Jeux de données de référence (dimensions)"]
        referentiel_pays["Référentiel des pays — EdStatsCountry.csv — PK : Country Code"]
        dictionnaire_indicateurs["Dictionnaire des indicateurs — EdStatsSeries.csv — PK : Series Code"]
    end
    subgraph faits["Jeux de données de mesures (faits)"]
        mesures["Mesures par pays et indicateur — EdStatsData.csv — PK : Country Code + Indicator Code"]
        sources["Sources par pays et indicateur — EdStatsCountry-Series.csv — PK : CountryCode + SeriesCode"]
        annotations["Annotations par pays, indicateur et année — EdStatsFootNote.csv — PK : CountryCode + SeriesCode + Year"]
    end
    mesures -->|"Country Code"| referentiel_pays
    mesures -->|"Indicator Code = Series Code"| dictionnaire_indicateurs
    sources -->|"CountryCode"| referentiel_pays
    sources -->|"SeriesCode"| dictionnaire_indicateurs
    annotations -->|"CountryCode"| referentiel_pays
    annotations -->|"SeriesCode"| dictionnaire_indicateurs
```
